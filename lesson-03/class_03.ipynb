```python
class Person:

    def __init__(self, name, age, number):
        self.name = name
        self._age = age
        self.__number = number

    def get_number(self):
        return self.__number

    def register(self):
        print(f"{self.name=}, {self._age=}, {self.__number=}")


class Student(Person):

    def __init__(self, name, age, number):
        super().__init__(name, age, number)
    
        self.name = "Student_" + name
        self._age = 1000 + age
        self.__number = "Student_" + number

    def register(self):
        super().register()
        print(f"Student: {self.get_number()=}, {self.__number=}")


steve = Student("Steve", 77, "112233")
```

```python
from abc import ABC, abstractmethod
```

```python
class Person(ABC):

    @abstractmethod
    def eat(self, food):
        print(f"Person.eat {food=}")

    @abstractmethod
    def sleep(self):
        return NotImplemented

    def study(self):
        self.sleep()
        self.eat("fruits")
```

```python
steve = Person()
```

```python
class Student(Person):
    def eat(self, food):
        print(f"Student.eat {food=}")
        super().eat(food)
```

```python
steve = Student()
```

```python
class BMSTUStudent(Student):
    def sleep(self):
        print("BMSTUStudent.sleep")
```

```python
steve = BMSTUStudent()
```

```python
class Predictable(ABC):
    @abstractmethod
    def predict(self):
        pass

    @classmethod
    def __subclasshook__(cls, C):
        print(f"{cls=}, {C=}")

        if cls is not Predictable:
            return NotImplemented

        name = "predict"
        for class_ in C.__mro__:
            if name in class_.__dict__ and callable(class_.__dict__[name]):
                return True

        return NotImplemented


class TreeModel:
    def predict(self, data):
        print("TreeModel", data)


class TrafficLight:
    predict = "green"


tree = TreeModel()
light = TrafficLight()


print(f"{isinstance(tree, Predictable)=}, {isinstance(light, Predictable)=}")
```

```python
class Predictable(ABC):
    @abstractmethod
    def predict(self):
        pass

    @classmethod
    def __subclasshook__(cls, C):
        #print(f"{cls=}, {C=}")

        if cls is not Predictable:
            return NotImplemented

        name = "predict"
        for class_ in C.__mro__:
            if name in class_.__dict__ and callable(class_.__dict__[name]):
                return True

        return NotImplemented


class TrainPredictable(Predictable):
    @abstractmethod
    def train(self):
        pass

    @classmethod
    def __subclasshook__(cls, C):
        #print(f"{cls=}, {C=}")

        if cls is not TrainPredictable:
            return NotImplemented

        for name in ("predict", "train"):
            for class_ in C.__mro__:
                if name in class_.__dict__ and callable(class_.__dict__[name]):
                    break
            else:
                return NotImplemented

        return True


class TreeModel:
    def predict(self, data):
        print("TreeModel", data)


class RandomForest:
    def predict(self, data):
        print("RandomForest", data)

    def train(self, data):
        print("RandomForest", data)


class TrafficLight:
    predict = "green"


tree = TreeModel()
forest = RandomForest()
light = TrafficLight()


print(f"{isinstance(tree, Predictable)=}, {isinstance(light, Predictable)=}, {isinstance(forest, Predictable)=}")
print(f"{isinstance(tree, TrainPredictable)=}, {isinstance(light, TrainPredictable)=}, {isinstance(forest, TrainPredictable)=}")
```
