```python
class Base:
    def __init__(self):
        self.__val = "base"

    @property
    def val(self):
        return self.__val


class Impl(Base):
    def __init__(self):
        super().__init__()
        self.__val = "impl"


data = Impl()
print(data.val)
```

```python
class Base:
    def __init__(self):
        self.__val = "base"

    @property
    def val(self):
        return self.__val


class Impl(Base):
    def __init__(self):
        super().__init__()
        self.__val = "impl"

    @property
    def val(self):
        return self.__val


data = Impl()
print(data.val)
```

```python
class Base:
    def __init__(self):
        self._val = "base"

    @property
    def val(self):
        print(f"proprt {self=}")
        return self._val


class Impl(Base):
    def __init__(self):
        super().__init__()
        self._val = "impl"


data = Impl()
print(f"before {data=}")
print(data.val)
```
