```python
import threading
import time
```

```python
%%time


N = 10 ** 9
N_THREADS = 10
N_JOBS = N // N_THREADS


def counter(a, b):
    print(f"name={threading.current_thread().name} {a=}, {b=} -- started")
    while a < b:
        a += 1
    print(f"name={threading.current_thread().name} -- finished")


def run():
    threads = [
        threading.Thread(
            target=counter,
            args=(i * N_JOBS, (i + 1) * N_JOBS),
            name=f"th_counter_{i}"
        )
        for i in range(N_THREADS)
    ]

    for th in threads:
        th.start()

    for th in threads:
        th.join()


run()
```

```python
%%time

N = 20

URL = "https://www.python.org/"
URLS = [URL] * N

N_THREADS = 10
N_JOBS = N // N_THREADS


def fetch_urls(urls, resps, sem):
    print(f"name={threading.current_thread().name} -- started")

    for url in urls:
        # print(requests.get(url).status_code)
        try:
            with sem:
                resp = requests.get(url)
            resps.append(resp.status_code)
        except Exception as err:
            print(f"ERROR: {err}")
            resps.append(-1)

    print(f"name={threading.current_thread().name} -- finished")


def run():
    resps = []
    sem = threading.Semaphore(10)

    threads = [
        threading.Thread(
            target=fetch_urls,
            args=(URLS[i * N_JOBS : (i + 1) * N_JOBS], resps, sem),
            name=f"th_fetch_{i}"
        )
        for i in range(N_THREADS)
    ]

    for th in threads:
        th.start()

    for th in threads:
        th.join()

    print(Counter(resps))


run()
```

```python
%%time

N = 20

URL = "https://www.python.org/"
URLS = [URL] * N

N_THREADS = 10
N_JOBS = N // N_THREADS


def fetch_urls_worker(que, resps):
    print(f"name={threading.current_thread().name} -- started")
    cnt = 0
    t1 = time.time()

    while True:
        url = que.get()
        if url is None:
            que.put(url)
            break

        cnt += 1
        try:
            resp = requests.get(url)
        except Exception as err:
            print(f"ERROR: {err}")
            resps.append(-1)
        else:
            resps.append(resp.status_code)

    t2 = time.time()

    print(f"name={threading.current_thread().name} -- finished", t2 - t1, cnt)


def run():
    resps = []
    que = queue.Queue()
    for url in URLS:
        que.put(url)
    que.put(None)

    threads = [
        threading.Thread(
            target=fetch_urls_worker,
            args=(que, resps),
            name=f"th_fetch_{i}"
        )
        for i in range(N_THREADS)
    ]

    for th in threads:
        th.start()

    for th in threads:
        th.join()

    print(Counter(resps))


run()
```
